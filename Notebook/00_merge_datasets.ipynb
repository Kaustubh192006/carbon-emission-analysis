```python
import pandas as pd
import os

data_folder = "../Data/raw"

# Read 2011 as our reference
df_2011 = pd.read_excel(
    os.path.join(data_folder, "ghgp_data_2011.xlsx"),
    sheet_name="Direct Emitters",
    header=3
)

reference_columns = set(df_2011.columns)

for year in range(2012, 2018):
    
    file_path = os.path.join(data_folder, f"ghgp_data_{year}.xlsx")
    
    df = pd.read_excel(
        file_path,
        sheet_name="Direct Emitters",
        header=3
    )
    
    current_columns = set(df.columns)
    
    missing = reference_columns - current_columns
    extra = current_columns - reference_columns
    
    print(f"\n{year}")
    print("Missing compared to 2011:", missing)
    print("Extra compared to 2011:", extra)
```

```python
import pandas as pd
import os

data_folder = "../Data/raw"

# Use 2018 as the reference for the Direct Point Emitters format
df_2018 = pd.read_excel(
    os.path.join(data_folder, "ghgp_data_2018.xlsx"),
    sheet_name="Direct Point Emitters",
    header=3
)

reference_columns = set(df_2018.columns)

for year in range(2019, 2024):

    file_path = os.path.join(data_folder, f"ghgp_data_{year}.xlsx")

    df = pd.read_excel(
        file_path,
        sheet_name="Direct Point Emitters",
        header=3
    )

    current_columns = set(df.columns)

    missing = reference_columns - current_columns
    extra = current_columns - reference_columns

    print(f"\n{year}")
    print("Missing compared to 2018:", missing)
    print("Extra compared to 2018:", extra)
```

```python
import pandas as pd
import os

data_folder = "../Data/raw"

all_years = []

for year in range(2011, 2024):

    file_path = os.path.join(
        data_folder,
        f"ghgp_data_{year}.xlsx"
    )

    # EPA changed the sheet name starting in 2018
    if year <= 2017:
        sheet_name = "Direct Emitters"
    else:
        sheet_name = "Direct Point Emitters"

    print(f"Reading {year}...")

    df = pd.read_excel(
        file_path,
        sheet_name=sheet_name,
        header=3
    )

    # Add the year
    df["Year"] = year

    all_years.append(df)

print("\nAll files loaded successfully.")
```

```python
master_df = pd.concat(
    all_years,
    ignore_index=True
)

print("Master dataset created.")
print("Rows:", master_df.shape[0])
print("Columns:", master_df.shape[1])
```

```python
print(master_df["Year"].unique())
```

```python
processed_folder = "../Data/processed"

os.makedirs(processed_folder, exist_ok=True)

output_file = os.path.join(
    processed_folder,
    "carbon_emission_master.csv"
)

master_df.to_csv(
    output_file,
    index=False
)

print("Saved successfully to:")
print(output_file)
```
